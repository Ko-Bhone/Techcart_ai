Test - Create, Read, Update, Delete

```python
import uuid
from src.database.connection import SessionLocal
from src.database.crud import create_product, get_product, list_products, update_product, delete_product

demo_product_id = (f"DEMO-{uuid.uuid4().hex[:8]}")

with SessionLocal() as session:

    #1. Create
    created = create_product(
        session,
        {
            "product_id": demo_product_id,
            "name" : "Demo NVMe SSD",
            "category" : "SSD",
            "brand" : "TechCart",
            "price" : 89.99,
            "stock" : 12,
            "rating" : 4.5
        }
    )

    print("Created product: ", created)

    #2. Read one Product
    found = get_product(session, demo_product_id)
    print("Read product: ", found)

    #3. Update
    updated = update_product(session, demo_product_id,
                             {
                                 "price":79.99,
                                 "stock": 20
                             })
    print("Updated product: ", updated)

    #4. List product by Category
    ssd_products = list_products(session, category="SSD", limit=10)
    print("SSD product counts: ", ssd_products)

    #5. Delete the demo product
    deleted = delete_product(session, demo_product_id)
    print("\nDeleted product: ", deleted)

    #Confirm deletion
    print("After deletion:", get_product(session, demo_product_id))
```
